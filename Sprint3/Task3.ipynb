**Описание задачи:**  
Создать базовый класс ***Book*** в C#, который будет представлять информацию о
книгах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
     
**Требования к базовому классу Book:**  
* Атрибуты: Название (***Title***), Автор (***Author***), Год издания (***YearOfPublication***).  
* Методы:  
    * ***GetInfo()***: метод для получения информации о книге в виде строки.  
    * ***Read()***: метод для вывода сообщения о чтении книги.  
    * ***Borrow()***: метод для вывода сообщения о выдаче книги на чтение.  
  
**Требования к производным классам:**
* Учебник (***Textbook***): Должен содержать дополнительные атрибуты, такие как
Предмет (***Subject***). Метод *Read()* должен быть переопределен для
добавления информации о предмете при чтении учебника.
* Художественная литература (***Fiction***): Должен содержать дополнительные
атрибуты, такие как Жанр (***Genre***). Метод *Borrow()* должен быть
переопределен для добавления информации о жанре при выдаче книги на
чтение.
* Научная литература (***ScientificLiterature***) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Область науки
(***FieldOfScience***). Метод *GetInfo()* должен быть переопределен для включения
информации об области науки в описании книги.

#### Дополнительное задание

```
//Интерфейс, который определяет методы для выдачи (Borrow) и возврата (Return) книги.
public interface IBorrowable
{
    void Borrow();
    void Borrow(int days); // Перегрузка метода Borrow
    void Return();
}

//Книги в целом (базовый класс)
public class Book : IBorrowable
{
    public string Title { get; set; } // Название книги
    public string Author { get; set; } // Автор книги
    public int YearOfPublication { get; set; } // Год издания
    public string ISBN { get; set; } // Уникальный идентификатор книги
    public string Publisher { get; set; } // Издательство
    public int NumberOfPages { get; set; } // Количество страниц
    public string Language { get; set; } // Язык книги
    public bool IsAvailable { get; set; } = true; // Доступность книги (по умолчанию true)

    private static int bookCount = 0; // Статическая переменная для подсчета количества книг

    // Конструктор класса
    public Book(string title, string author, int yearOfPublication, string isbn, string publisher, int numberOfPages, string language)
    {
        Title = title;
        Author = author;
        YearOfPublication = yearOfPublication;
        ISBN = isbn;
        Publisher = publisher;
        NumberOfPages = numberOfPages;
        Language = language;
        bookCount++; // Увеличиваем счетчик книг при создании нового объекта
    }

    // Статический метод для получения количества книг
    public static int GetBookCount()
    {
        return bookCount;
    }

    // Метод для получения информации о книге
    public virtual string GetInfo()
    {
        return $"Название: {Title}, Автор: {Author}, Год издания: {YearOfPublication}, ISBN: {ISBN}, Издательство: {Publisher}, Страниц: {NumberOfPages}, Язык: {Language}, Доступна: {(IsAvailable ? "Да" : "Нет")}";
    }

    // Метод для чтения книги
    public virtual void Read()
    {
        Console.WriteLine($"Чтение книги: {Title} (Автор: {Author}, Год издания: {YearOfPublication})");
    }

    // Метод для выдачи книги
    public virtual void Borrow()
    {
        if (IsAvailable)
        {
            Console.WriteLine($"Книга '{Title}' выдана на чтение.");
            IsAvailable = false;
        }
        else
        {
            Console.WriteLine($"Книга '{Title}' недоступна для выдачи.");
        }
    }

    // Перегрузка метода Borrow (реализация интерфейса IBorrowable)
    public virtual void Borrow(int days)
    {
        if (IsAvailable)
        {
            Console.WriteLine($"Книга '{Title}' выдана на {days} дней.");
            IsAvailable = false;
        }
        else
        {
            Console.WriteLine($"Книга '{Title}' недоступна для выдачи.");
        }
    }

    // Метод для возврата книги
    public virtual void Return()
    {
        Console.WriteLine($"Книга '{Title}' возвращена.");
        IsAvailable = true;
    }

    // Метод для проверки доступности книги
    public void CheckAvailability()
    {
        Console.WriteLine($"Книга '{Title}' доступна: {(IsAvailable ? "Да" : "Нет")}");
    }
}

//Учебники
public class Textbook : Book
{
    // Дополнительные свойства для учебника
    public string Subject { get; set; } // Предмет учебника
    public int Edition { get; set; } // Издание учебника
    public bool IsRecommended { get; set; } // Рекомендован ли учебник

    // Конструктор класса
    public Textbook(string title, string author, int yearOfPublication, string isbn, string publisher, int numberOfPages, string language, string subject, int edition, bool isRecommended)
        : base(title, author, yearOfPublication, isbn, publisher, numberOfPages, language)
    {
        Subject = subject;
        Edition = edition;
        IsRecommended = isRecommended;
    }

    // Переопределение метода GetInfo() для добавления информации о предмете и издании
    public override string GetInfo()
    {
        return base.GetInfo() + $", Предмет: {Subject}, Издание: {Edition}, Рекомендован: {(IsRecommended ? "Да" : "Нет")}";
    }

    // Переопределение метода Read() для учебника
    public override void Read()
    {
        Console.WriteLine($"Чтение учебника по предмету {Subject}: {Title} (Издание: {Edition})");
    }

    // Метод для проверки рекомендательного статуса учебника
    public void CheckRecommendation()
    {
        Console.WriteLine($"Учебник '{Title}' рекомендован: {(IsRecommended ? "Да" : "Нет")}");
    }
    
    // Перекрытие метода Borrow (сокрытие базовой реализации)
    public new void Borrow()
    {
        if (IsAvailable)
        {
            Console.WriteLine($"Учебник '{Title}' по предмету {Subject} выдан на стандартный срок.");
            IsAvailable = false;
        }
        else
        {
            Console.WriteLine($"Учебник '{Title}' по предмету {Subject} недоступен для выдачи.");
        }
    }

    // Перегрузка метода Borrow (дополнительный параметр)
    public void Borrow(string borrowerType)
    {
        if (IsAvailable)
        {
            Console.WriteLine($"Учебник '{Title}' выдан для {borrowerType} на стандартный срок.");
            IsAvailable = false;
        }
        else
        {
            Console.WriteLine($"Учебник '{Title}' недоступен для выдачи.");
        }
    }

}

//Художественная литератра
public class Fiction : Book
{
    // Дополнительные свойства для художественной литературы
    public string Genre { get; set; } // Жанр книги
    public string Awards { get; set; } // Награды книги
    public bool IsBestseller { get; set; } // Является ли книга бестселлером

    // Конструктор класса
    public Fiction(string title, string author, int yearOfPublication, string isbn, string publisher, int numberOfPages, string language, string genre, string awards, bool isBestseller)
        : base(title, author, yearOfPublication, isbn, publisher, numberOfPages, language)
    {
        Genre = genre;
        Awards = awards;
        IsBestseller = isBestseller;
    }

    // Переопределение метода GetInfo() для добавления информации о жанре и наградах
    public override string GetInfo()
    {
        return base.GetInfo() + $", Жанр: {Genre}, Награды: {Awards}, Бестселлер: {(IsBestseller ? "Да" : "Нет")}";
    }

    // Перекрытие метода Read 
    public override void Read()
    {
        base.Read(); // Вызов базовой реализации
        Console.WriteLine($"Погружение в мир жанра {Genre}...");
    }

    // Перегрузка метода Read (дополнительный параметр)
    public void Read(string readingMode)
    {
        Console.WriteLine($"Чтение книги '{Title}' в режиме {readingMode}.");
    }

    // Переопределение метода Borrow() для художественной литературы
    public override void Borrow()
    {
        if (IsAvailable)
        {
            Console.WriteLine($"Книга '{Title}' (Жанр: {Genre}) выдана на чтение.");
            IsAvailable = false;
        }
        else
        {
            Console.WriteLine($"Книга '{Title}' (Жанр: {Genre}) недоступна для выдачи.");
        }
    }

    // Метод для проверки статуса бестселлера
    public void CheckBestsellerStatus()
    {
        Console.WriteLine($"Книга '{Title}' является бестселлером: {(IsBestseller ? "Да" : "Нет")}");
    }
}

//Научная литература
public class ScientificLiterature : Book
{
    // Дополнительные свойства для научной литературы
    public string FieldOfScience { get; set; } // Область науки
    public int Citations { get; set; } // Количество цитирований
    public bool IsPeerReviewed { get; set; } // Рецензирована ли книга

    // Конструктор класса
    public ScientificLiterature(string title, string author, int yearOfPublication,string isbn, string publisher, int numberOfPages, string language, string fieldOfScience, int citations, bool isPeerReviewed)
        : base(title, author, yearOfPublication, isbn, publisher, numberOfPages, language) 
    {
        FieldOfScience = fieldOfScience;
        Citations = citations;
        IsPeerReviewed = isPeerReviewed;
    }

    // Переопределение метода GetInfo() для добавления информации о научной области и цитированиях
    public override string GetInfo()
    {
        return base.GetInfo() + $", Область науки: {FieldOfScience}, Цитирования: {Citations}, Рецензирована: {(IsPeerReviewed ? "Да" : "Нет")}";
    }

    // Метод для проверки статуса рецензирования
    public void CheckPeerReviewStatus()
    {
        Console.WriteLine($"Научная литература '{Title}' рецензирована: {(IsPeerReviewed ? "Да" : "Нет")}");
    }

    // Перегрузка метода Return (добавление параметра)
    public void Return(string comments)
    {
        Console.WriteLine($"Книга '{Title}' возвращена с комментарием: '{comments}'");
        IsAvailable = true;
    }
}

//Продвинутые учебники
public class AdvancedTextbook : Textbook
{
    // Дополнительные свойства для продвинутого учебника
    public string DifficultyLevel { get; set; } // Уровень сложности
    public bool IsDigital { get; set; } // Доступна ли цифровая версия

     // Конструктор класса
    public AdvancedTextbook(string title, string author, int yearOfPublication, string isbn, string publisher, int numberOfPages, string language, string subject, int edition, bool isRecommended, string difficultyLevel, bool isDigital)
        : base(title, author, yearOfPublication, isbn, publisher, numberOfPages, language, subject, edition, isRecommended)
    {
        DifficultyLevel = difficultyLevel;
        IsDigital = isDigital;
    }

    // Переопределение метода GetInfo() для добавления информации об уровне сложности и цифровой версии
    public override string GetInfo()
    {
        return base.GetInfo() + $", Уровень сложности: {DifficultyLevel}, Цифровая версия: {(IsDigital ? "Да" : "Нет")}";
    }

    // Перекрытие метода Read
    public override void Read()
    {
        Console.WriteLine($"Изучение продвинутого учебника '{Title}' (Уровень: {DifficultyLevel})");
    }

    // Метод для проверки доступности цифровой версии
    public void CheckDigitalAvailability()
    {
        Console.WriteLine($"Учебник '{Title}' доступен в цифровом формате: {(IsDigital ? "Да" : "Нет")}");
    }
    
    // Перегрузка метода CheckDigitalAvailability
    public void CheckDigitalAvailability(bool includePrice)
    {
        Console.WriteLine($"Учебник '{Title}' доступен в цифровом формате: {(IsDigital ? "Да" : "Нет")}");
        if (includePrice && IsDigital)
        {
            Console.WriteLine("Цена цифровой версии: 300₽");
        }
    }

}

//Историческая художественная литература
public class HistoricalFiction : Fiction
{
    // Дополнительные свойства для исторической художественной литературы
    public string HistoricalPeriod { get; set; } // Исторический период
    public string HistoricalAccuracy { get; set; } // Историческая точность
    public string MainCharacter { get; set; } // Главный герой

    // Конструктор класса
    public HistoricalFiction(string title, string author, int yearOfPublication, string isbn, string publisher, int numberOfPages, string language, string genre, string awards, bool isBestseller, string historicalPeriod, string historicalAccuracy, string mainCharacter)
        : base(title, author, yearOfPublication, isbn, publisher, numberOfPages, language, genre, awards, isBestseller)
    {
        HistoricalPeriod = historicalPeriod;
        HistoricalAccuracy = historicalAccuracy;
        MainCharacter = mainCharacter;
    }

    // Переопределение метода GetInfo() для добавления информации о историческом периоде и главном герое
    public override string GetInfo()
    {
        return base.GetInfo() + $", Исторический период: {HistoricalPeriod}, Историческая точность: {HistoricalAccuracy}, Главный герой: {MainCharacter}";
    }

    // Метод для проверки исторической точности
    public void CheckHistoricalAccuracy()
    {
        Console.WriteLine($"Историческая точность книги '{Title}': {HistoricalAccuracy}");
    }

     // Перегрузка метода CheckBestsellerStatus
    public void CheckBestsellerStatus(string region)
    {
        Console.WriteLine($"Книга '{Title}' является бестселлером в {region}: {(IsBestseller ? "Да" : "Нет")}");
    }
}

// Generic класс для управления коллекцией книг
public class Library<T> where T : Book
{
    private List<T> books = new List<T>();

    // Добавление книги в коллекцию
    public void AddBook(T book)
    {
        books.Add(book);
        Console.WriteLine($"Книга '{book.Title}' добавлена в коллекцию.");
    }

    // Удаление книги из коллекции
    public void RemoveBook(T book)
    {
        if (books.Remove(book))
        {
            Console.WriteLine($"Книга '{book.Title}' удалена из коллекции.");
        }
        else
        {
            Console.WriteLine($"Книга '{book.Title}' не найдена в коллекции.");
        }
    }

    // Поиск книги по названию
    public T FindBookByTitle(string title)
    {
        return books.FirstOrDefault(b => b.Title.Equals(title, StringComparison.OrdinalIgnoreCase));
    }

    // Вывод информации о всех книгах в коллекции
    public void DisplayAllBooks()
    {
        Console.WriteLine($"\nКоллекция содержит {books.Count} книг:");
        foreach (var book in books)
        {
            Console.WriteLine(book.GetInfo());
        }
    }
}

public class BookCatalog<T> where T : Book
{
    private Dictionary<string, T> catalog = new Dictionary<string, T>();

    public void AddToCatalog(T book)
    {
        catalog[book.ISBN] = book;
        Console.WriteLine($"Книга '{book.Title}' добавлена в каталог.");
    }

    public T FindByISBN(string isbn)
    {
        catalog.TryGetValue(isbn, out T book);
        return book;
    }

    public void DisplayCatalog()
    {
        Console.WriteLine("\nКаталог книг:");
        foreach (var book in catalog.Values)
        {
            Console.WriteLine(book.GetInfo());
        }
    }
}
```

```
Console.WriteLine("=== Демонстрация полиморфизма ===");

var book = new Book("Основы программирования", "Иванов И.И.", 2020, 
                    "123-456-789", "Техносфера", 350, "Русский");

var textbook = new Textbook("Алгебра и начала анализа", "Петров П.П.", 2019, 
                            "987-654-321", "Просвещение", 200, "Русский", 
                            "Математика", 5, true);

var fiction = new Fiction("1984", "Джордж Оруэлл", 1949, 
                            "111-222-333", "АСТ", 328, "Русский", 
                            "Антиутопия", "Премия Хьюго", true);

var sciLit = new ScientificLiterature("Краткая история времени", "Стивен Хокинг", 1988, 
                                    "444-555-666", "Мир", 256, "Русский", 
                                    "Космология", 15000, true);

var advTextbook = new AdvancedTextbook("Продвинутая алгебра", "Сидоров С.С.", 2021, 
                                        "777-888-999", "Физматлит", 450, "Русский", 
                                        "Математика", 3, true, "Эксперт", true);

var histFiction = new HistoricalFiction("Война и мир", "Лев Толстой", 1869, 
                                        "000-111-222", "Эксмо", 1225, "Русский", 
                                        "Роман-эпопея", "Нобелевская премия", true, 
                                        "Эпоха Наполеона", "Высокая", "Пьер Безухов");


Console.WriteLine("\n=== Перегрузка методов ===");

// Перегруженный метод Borrow() с параметром
book.Borrow(30); // Вызов Borrow(int days) из Book

// Перегруженный метод Borrow() с другим параметром
textbook.Borrow("студента"); // Вызов Borrow(string borrowerType) из Textbook

// Перегруженный метод Return() с параметром
sciLit.Return("Отличное состояние"); // Вызов Return(string comments) из ScientificLiterature

// Перегруженный метод Read() с параметром
fiction.Read("аудио"); // Вызов Read(string readingMode) из Fiction

// Перегруженный метод CheckDigitalAvailability с параметром
advTextbook.CheckDigitalAvailability(true); // Вызов CheckDigitalAvailability(bool includePrice) из AdvancedTextbook

// Перегруженный метод CheckBestsellerStatus с параметром
histFiction.CheckBestsellerStatus("Европе"); // Вызов CheckBestsellerStatus(string region) из HistoricalFiction


Console.WriteLine("\n=== Перекрытие методов ===");

// Перекрытие метода Borrow() - вызовется метод из Textbook, а не из Book
textbook.Borrow(); // Вызов new Borrow() из Textbook

// Переопределение метода Read() - вызовется метод из Fiction, а не из Book
fiction.Read(); // Вызов override Read() из Fiction

// Переопределение метода Read() - вызовется метод из AdvancedTextbook
advTextbook.Read(); // Вызов override Read() из AdvancedTextbook

Console.WriteLine(histFiction.GetInfo()); // Вызов override GetInfo() из HistoricalFiction


Console.WriteLine("\n=== Работа generic-классов ===");

// Создаем библиотеку для книг
Library<Book> bookLibrary = new Library<Book>();
bookLibrary.AddBook(book);
bookLibrary.AddBook(textbook); // Textbook наследуется от Book
bookLibrary.DisplayAllBooks();
Console.WriteLine("\n");

// Создаем библиотеку для учебников
Library<Textbook> textbookLibrary = new Library<Textbook>();
textbookLibrary.AddBook(textbook);
textbookLibrary.AddBook(advTextbook); // AdvancedTextbook наследуется от Textbook
textbookLibrary.DisplayAllBooks();
Console.WriteLine("\n");

// Создаем каталог для художественной литературы
BookCatalog<Fiction> fictionCatalog = new BookCatalog<Fiction>();
fictionCatalog.AddToCatalog(fiction);
fictionCatalog.AddToCatalog(histFiction); // HistoricalFiction наследуется от Fiction
fictionCatalog.DisplayCatalog();
Console.WriteLine("\n");

// Поиск книги в каталоге
var foundBook = fictionCatalog.FindByISBN("111-222-333");
if (foundBook != null)
{
    Console.WriteLine($"\nНайдена книга: {foundBook.Title}");
}
```
